### Model Integration with Google Gemeni and Groq

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model=init_chat_model(
    "gemini-3.1-flash-lite",
     model_provider="google_genai",
     thinking_budget=0,
     )
model
```

```python
# invoke model
response= model.invoke("Why do parrot talk")
response.content
```

### Another way of Invoking based on the model we are using 

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(
    model="gemini-3.1-flash-lite"
)
response = model.invoke("What is you name?")
response.content
```

### Streaming and Batch

### Streaming

When ussing the invoke we have to wait untill the complete output is generated, therefore we can use the streaming to stream the output while it is being generated. And as the stream returns a generator we need a loop to print the content of it. 

```python
for chunk in model.stream("Write me a 200 words paragraph on Police"):
    print(chunk.text)
```

### Batch

It is a collection of independed input for the model which can be given to the model parallaly to improve the performance and reduce the cost.

```python
responses = model.batch([
    "Why parrot have colourfull feather",
    "why sky is blue",
    "Why humans die",],
    config={
        'max_concurrency': 1, # limit the number of parallel calls
    }
)
for response in response:
    print(response)
```
